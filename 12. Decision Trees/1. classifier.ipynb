```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
iris
```

```python
X = pd.DataFrame(iris['data'], columns=['sepal length','sepal width','petal length','petal width'])
```

```python
# Series
y = iris['target']
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=23)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree_classifier = DecisionTreeClassifier()
```

```python
tree_classifier.fit(X_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(tree_classifier, filled=True)
```

```python
y_pred = tree_classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_pred,y_test))
print(accuracy_score(y_pred,y_test))
print(classification_report(y_pred,y_test))
```

```python

# ---------------------------------------------------------------------------------
# What each hyperparameter means (beginner-friendly)
# ---------------------------------------------------------------------------------
# Think of a Decision Tree as a game of "20 Questions".
# The tree keeps asking yes/no questions about your data until it can guess the answer.
#
# criterion             -> How the tree decides which question is the "best" one to ask.
#                          'gini', 'entropy' and 'log_loss' are just different formulas
#                          for measuring how "mixed up" the groups are. The goal is groups
#                          that are as pure as possible (all the same class).
#
# splitter              -> How the tree picks the question at each step.
#                          'best'   = checks all options and picks the best one (slower, more careful)
#                          'random' = picks from a few random options (faster, adds variety)
#
# max_depth             -> The most questions the tree can ask in a row (how tall it can grow).
#                          Small number = simple tree (may miss patterns -> underfitting)
#                          Large / None = very detailed tree (may memorize data -> overfitting)
#
# min_samples_split     -> The fewest data points a group needs before the tree is
#                          allowed to split it into two smaller groups.
#                          Higher value = fewer splits = simpler tree.
#
# min_samples_leaf      -> The fewest data points allowed in a final group (a "leaf").
#                          Stops the tree from making tiny groups with just 1 data point.
#
# max_features          -> How many columns (features) the tree looks at when picking a question.
#                          None   = look at all features
#                          'sqrt' = look at the square root of the number of features
#                          'log2' = look at log2 of the number of features
#                          Looking at fewer features adds randomness and can reduce overfitting.
#
# max_leaf_nodes        -> The most final groups (leaves) the tree can have.
#                          None = no limit. A smaller number keeps the tree small and simple.
#
# min_impurity_decrease -> A split only happens if it makes the groups at least this much "purer".
#                          Stops the tree from making splits that barely help.
#
# class_weight          -> How much each class matters.
#                          None       = every class counts the same
#                          'balanced' = gives more importance to rare classes
#                          (useful when one class has far more rows than another)
#
# ccp_alpha             -> "Pruning": cuts off branches that don't help much after the tree is built.
#                          0.0 = no pruning. Bigger value = more branches removed = simpler tree.
# ---------------------------------------------------------------------------------
```
