## DBSCAN Clustering

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### Create a moon shaped dataset

```python
from sklearn.datasets import make_moons

# 2 half-moon shapes with a little noise
X, y = make_moons(n_samples=250, noise=0.05, random_state=42)
X[:5]
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.title("Raw Data")
plt.show()
```

### Scale the features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

### Apply DBSCAN

```python
from sklearn.cluster import DBSCAN

# eps = neighbourhood radius, min_samples = points needed to form a dense region
model = DBSCAN(eps=0.3, min_samples=5)
labels = model.fit_predict(X_scaled)
labels  # -1 means noise / outlier
```

### Compare predicted clusters with actual labels

```python
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.title("DBSCAN Clusters")
plt.show()
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.title("Actual Labels")
plt.show()
```
