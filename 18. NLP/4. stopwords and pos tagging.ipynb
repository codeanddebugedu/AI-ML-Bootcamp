## Stopwords
Stopwords are very common words like **is, the, a, and, in, of** that appear in almost every sentence but add little meaning.

Removing them keeps only the important words, so our data becomes smaller and the model focuses on what matters.

Example: *"This is a very good phone"* → **good, phone**

### Real life use case
**Search engine:** you search *"what is the best laptop for students"*. Words like *what, is, the, for* appear on millions of pages, so they don't help. The search engine removes them and searches using **best, laptop, students**, which gives much better results.

### Remove stopwords from a sentence

```python
from nltk.tokenize import word_tokenize

# set makes checking 'word in stop_words' faster
stop_words = set(stopwords.words('english'))

sentence = "This is a very good phone and I am happy with it"

# lower() because the stopwords list is in small letters ('This' would not match 'this')
words = word_tokenize(sentence.lower())
print(words)
filtered = [word for word in words if word not in stop_words]
filtered
```

```python
# be careful: 'not' is also a stopword, so the meaning flips!
sentence = "This phone is not good"
[word for word in word_tokenize(sentence.lower()) if word not in stop_words]
```

### Real example: paragraph → stopwords removal → stemming / lemmatization
This is how text is cleaned before giving it to an ML model.

```python
paragraph = """Artificial Intelligence is changing the way we live and work. Today, AI is being used in hospitals to detect diseases early.
Banks are using it to catch fraud before it happens. Students are learning faster with the help of AI tutors.
But AI is not magic, it learns from the data we give it. If the data is wrong, the results will also be wrong.
That is why learning the basics of data and machine learning is very important.
In this bootcamp, we are going to build these skills step by step."""
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(paragraph)
sentences
```

### Stopwords + Porter Stemmer

```python
from nltk.stem import PorterStemmer

porter = PorterStemmer()

porter_sentences = []
for sentence in sentences:
    words = word_tokenize(sentence.lower())
    words = [porter.stem(word) for word in words if word not in stop_words]
    porter_sentences.append(" ".join(words))  # join words back into a sentence

porter_sentences
```

### Stopwords + Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

snowball = SnowballStemmer('english')

snowball_sentences = []
for sentence in sentences:
    words = word_tokenize(sentence.lower())
    words = [snowball.stem(word) for word in words if word not in stop_words]
    snowball_sentences.append(" ".join(words))

snowball_sentences
```

### Stopwords + Lemmatizer

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemma_sentences = []
for sentence in sentences:
    words = word_tokenize(sentence.lower())
    # pos='v' treats every word as a verb (to keep it simple)
    words = [lemmatizer.lemmatize(word, pos='v') for word in words if word not in stop_words]
    lemma_sentences.append(" ".join(words))

lemma_sentences
```

### Stopwords + Lemmatizer with auto POS
Above we treated every word as a verb. Now let `pos_tag` find the correct POS of each word (like we did in the lemmatization notebook).

```python
# # needed to find the POS of each word
# nltk.download('averaged_perceptron_tagger_eng')
```

```python
# convert pos_tag's tag into lemmatizer's pos
# we only check the first letter, because all verb tags start with V (VB, VBD, VBG...),
# all adjective tags start with J (JJ, JJR, JJS) and all adverb tags start with R (RB, RBR, RBS)
def get_pos(tag:str) -> str:
    if tag.startswith("V"):
        return "v"  # verb
    elif tag.startswith("J"):
        return "a"  # adjective
    elif tag.startswith("R"):
        return "r"  # adverb
    else:
        return "n"  # noun (default for everything else)
```

```python
pos_sentences = []
for sentence in sentences:
    # find POS on the full sentence first (stopwords help pos_tag understand the sentence)
    tags = nltk.pos_tag(word_tokenize(sentence.lower()))
    words = [lemmatizer.lemmatize(word, pos=get_pos(tag)) for word, tag in tags if word not in stop_words]
    pos_sentences.append(" ".join(words))

pos_sentences
```

## POS Tagging (Part of Speech Tagging)
POS tagging means finding the **role of each word** in a sentence: is it a noun, verb, adjective, adverb...?

Example: *"Rahul is quickly writing a program"*
- Rahul → proper noun (name)
- is, writing → verb
- quickly → adverb
- program → noun

### Real life use case
- **Grammar checkers (like Grammarly):** to find mistakes, they first need to know which word is a verb, which is a noun and so on.
- **Voice assistants (like Alexa):** in *"set an alarm"*, **set** is a verb, so it knows what action to do and on what (**alarm**, a noun).
- **Lemmatization:** as we saw above, the lemmatizer needs the POS to give the correct base word.

```python
from nltk.tokenize import word_tokenize

sentence = "Rahul is quickly writing a beautiful program for his new company"

# pos_tag gives a (word, tag) pair for every word
tags = nltk.pos_tag(word_tokenize(sentence))
for word, tag in tags:
    print(f"{word} ----> {tag}")
```

### Common POS tags
NLTK uses the **Penn Treebank** tag list. You don't need to memorise all of them, just remember the first letters: **N**oun, **V**erb, **J** = adjective, **R** = adverb.

| Tag | Meaning | Example |
|---|---|---|
| `NN` | noun, singular | phone, city |
| `NNS` | noun, plural | phones, friends |
| `NNP` | proper noun (name) | Rahul, Mumbai |
| `NNPS` | proper noun, plural | Indians |
| `PRP` | pronoun | I, she, them |
| `PRP$` | possessive pronoun | my, his |
| `VB` | verb, base form | go, run |
| `VBD` | verb, past tense | went, bought |
| `VBG` | verb, -ing form | writing, reading |
| `VBN` | verb, past participle | eaten, written |
| `VBP` | verb, present (I / you / they) | play, watch |
| `VBZ` | verb, present (he / she / it) | runs, is |
| `JJ` | adjective | good, beautiful |
| `JJR` | adjective, comparative | bigger, older |
| `JJS` | adjective, superlative | biggest, fastest |
| `RB` | adverb | quickly, very |
| `RBR` | adverb, comparative | faster |
| `RBS` | adverb, superlative | most |
| `DT` | determiner | the, a, this |
| `IN` | preposition | in, for, than |
| `CC` | conjunction | and, but |
| `CD` | number | two, 2026 |
| `MD` | modal verb | will, can |
| `TO` | the word "to" | to |
