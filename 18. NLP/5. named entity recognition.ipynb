## Named Entity Recognition (NER)
NER finds the **important names** in a sentence and tells us **what type** each one is: a person, a company, a place...

Example: *"Sundar Pichai is the CEO of Google"*
- Sundar Pichai → **PERSON**
- Google → **ORGANIZATION**

### Common entity types
| Entity | Example |
|---|---|
| Person | Virat Kohli, Sundar Pichai |
| Organization | Google, Tata Motors |
| Place / Location | India, Mumbai, Himalayas |
| Date | 15 August 1947, 26-01-2025 |
| Time | 10:30am |
| Money | 5 lakh rupees, 1 million dollars |
| Percent | 18%, eighteen percent |

### Real life use case
- **Resume screening:** HR software reads thousands of resumes and pulls out the **name, companies worked at and city** of each candidate automatically.
- **News apps:** when you read an article, apps like Google News find the **people, companies and places** in it and show related stories about them.
- **Customer support:** in *"My order from Flipkart did not reach Pune"*, NER finds **Flipkart** (company) and **Pune** (place), so the ticket goes to the right team.

### Steps for NER in NLTK
1. **Tokenize** the sentence into words
2. **POS tag** every word (NER uses the tags, e.g. names are usually `NNP`)
3. **ne_chunk** groups the words into entities and gives each one a label

### Labels NLTK gives
| Label | Meaning | Example |
|---|---|---|
| `PERSON` | people | Sundar Pichai |
| `ORGANIZATION` | companies, teams, institutes | Reliance, ISRO |
| `GPE` | countries, states, cities | India, Mumbai |
| `LOCATION` | other places (rivers, mountains) | Himalayas |
| `FACILITY` | buildings, airports, bridges | Taj Mahal |

Note: NLTK's NER does **not** find dates, time, money or percent. Libraries like **spaCy** can find those too.
