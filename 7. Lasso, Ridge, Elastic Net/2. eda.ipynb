```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
df=pd.read_csv("forest_fire_dataset_cleaned.csv")
df = df.drop(['day', 'month', 'year'], axis=1)
```

```python
df["Classes"].value_counts()
```

```python
df["Classes"] = np.where(df["Classes"].str.contains("not"),0, 1)
```

```python
df.tail()
```

```python
df["Classes"].value_counts()
```

```python
plt.style.use('seaborn-v0_8')
df.hist(bins=50,figsize=(20,16))
plt.show()
```

```python
# Pie chart of classes
percentages=df["Classes"].value_counts(normalize=True) * 100
class_labels=["Fire","Not Fire"]
```

```python
plt.pie(percentages,labels=class_labels, autopct="%1.1f%%")
plt.title("Pie chart of Classes")
```
