## SVC

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
# We will create synthetic datapoints
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=1000, n_features=2, n_classes=2, 
                          n_clusters_per_class=1, 
                          n_redundant=0)
```

```python
pd.DataFrame(X)
```

```python
pd.DataFrame(y)
```

```python
sns.scatterplot(x=X[:,0], y=X[:,1], hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=22)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='linear')
```

```python
svc.fit(X_train,y_train)
```

```python
svc.coef_
```

```python
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_true=y_test,y_pred=y_pred))
print(confusion_matrix(y_true=y_test,y_pred=y_pred))
```

```python
rbf = SVC(kernel='rbf')
rbf.fit(X_train,y_train)
y_pred1 = rbf.predict(X_test)
print(classification_report(y_true=y_test,y_pred=y_pred1))
print(confusion_matrix(y_true=y_test,y_pred=y_pred1))
```

```python
poly = SVC(kernel='poly')
poly.fit(X_train,y_train)
y_pred2 = poly.predict(X_test)
print(classification_report(y_true=y_test,y_pred=y_pred2))
print(confusion_matrix(y_true=y_test,y_pred=y_pred2))
```

```python
sigmoid = SVC(kernel='sigmoid')
sigmoid.fit(X_train,y_train)
y_pred3 = sigmoid.predict(X_test)
print(classification_report(y_true=y_test,y_pred=y_pred3))
print(confusion_matrix(y_true=y_test,y_pred=y_pred3))
```

# Hypertuning SVC

```python
from sklearn.model_selection import GridSearchCV

# define parameter range
param_grid={
    "C":[0.1,1,10,100,1000],
    "gamma":[1,0.1,0.01,0.001,0.0001],
    "kernel":["rbf","linear","poly","sigmoid"]
}
```

```python
grid= GridSearchCV(SVC(),param_grid=param_grid,cv=5,verbose=3)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
print(classification_report(y_true=y_test,y_pred=y_pred))
print(confusion_matrix(y_true=y_test,y_pred=y_pred))
```
