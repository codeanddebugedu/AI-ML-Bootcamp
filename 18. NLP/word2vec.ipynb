## Word2Vec
In Bag of Words and TF-IDF, every word is just a column. The model has no idea that *"happy"* and *"glad"* mean almost the same thing.

Word2Vec converts every word into a **vector** (a list of numbers, here 300 numbers). Words with similar meaning get **similar vectors**, so they are close to each other.

- *doctor* is close to *nurse*, *hospital*
- *happy* is close to *glad*, *pleased*

It even learns relations, like **king - man + woman ≈ queen**.

### Real life use case
**Search and recommendations:** you search *"cheap mobile"* on a shopping site. With Word2Vec it knows *cheap* ≈ *affordable, budget* and *mobile* ≈ *phone, smartphone*, so it also shows products titled *"Budget Smartphone"*, even though none of your words are in the title.

**Chatbots:** a customer writes *"I want my money back"* or *"please refund"*. The words are different but the vectors are close, so the bot understands both mean **refund**.

### Pretrained model: Google News
We don't train our own model here. We use a model that **Google already trained** on about 100 billion words from Google News articles.

- It has **3 million** words and phrases
- Every word is a vector of **300** numbers

Note: the download is about **1.6 GB**, so it takes time the first time. After that it is loaded from your computer.

`gensim` does not install on Python 3.14 yet, so use **Python 3.12** (or Google Colab) for this notebook.

```python
# !pip install gensim
```

```python
import gensim.downloader as api

# downloads the model the first time (~1.6 GB), then loads it from your computer
wv = api.load('word2vec-google-news-300')
```

### Vector of a word

```python
vec_laptop = wv['laptop']
vec_laptop
```

```python
# every word is a list of 300 numbers
vec_laptop.shape
```

```python
# words are case sensitive: 'India' and 'india' are 2 different words
'India' in wv, 'india' in wv
```

```python
# some very common words like 'and', 'of', 'a' were removed by Google, so they are not in the model
# asking wv['and'] would give a KeyError
'and' in wv
```

### Most similar words
`most_similar` gives the 10 words whose vectors are closest to our word, with a score (closer to 1 = more similar).

```python
wv.most_similar('doctor')
```

```python
wv.most_similar('delicious')
```

```python
# phrases are joined with '_' in this model
wv.most_similar('Virat_Kohli')
```

### Similarity between 2 words
Score close to **1** = very similar, close to **0** = not related.

```python
wv.similarity('doctor', 'hospital')
```

```python
# not related, so the score should be much lower
wv.similarity('doctor', 'pizza')
```

### Odd one out

```python
# 'laptop' is not a meal
wv.doesnt_match(['breakfast', 'lunch', 'dinner', 'laptop'])
```

### Word maths: king - man + woman = ?
Take *king*, remove the "man" part and add the "woman" part. The closest word should be **queen**.

```python
vec = wv['king'] - wv['man'] + wv['woman']
vec
```

```python
# 'king' itself may come first, because our vector is still very close to 'king'
wv.most_similar([vec])
```

```python
# better way: positive words are added, negative words are subtracted
# this also removes the input words (king, man, woman) from the answer
wv.most_similar(positive=['king', 'woman'], negative=['man'])
```

```python
# same idea with countries and capitals: Paris - France + India = ?
wv.most_similar(positive=['Paris', 'India'], negative=['France'])
```
