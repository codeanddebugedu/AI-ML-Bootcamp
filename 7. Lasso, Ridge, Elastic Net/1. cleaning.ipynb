# Algerian Forest Fires - Regularised Regression

**What this notebook does.** Takes the raw Algerian forest fire records, repairs the
structural problems in the file, explores what separates a fire day from a quiet one,
and leaves behind a clean table ready for the Ridge and Lasso models that follow.

**Where the data comes from.** Daily field observations from two Algerian provinces,
collected across the 2012 fire season (June through September):

- **Bejaia** - north-east of the country
- **Sidi Bel-abbes** - north-west

**Shape of it.** 244 observations in total, split evenly at 122 per province. Each row
holds 11 predictors - four raw weather readings and six indices from the Canadian Fire
Weather Index system - plus a `Classes` label recording whether a fire broke out that
day. The label leans slightly toward fire: 138 fire days against 106 quiet ones.

**Why the cleaning comes first.** The two provinces are stacked one on top of the other
in a single file, with a title line above the real header and a second header buried in
the middle where the next block begins. Every column therefore arrives as text. None of
the analysis below works until the regions are tagged, the stray rows dropped, and the
dtypes restored.

### Columns

**Date**

| Column | Meaning |
| --- | --- |
| `day`, `month`, `year` | Calendar date of the observation. Months run June to September; the year is always 2012. |

**Weather readings taken at noon**

| Column | Meaning | Range |
| --- | --- | --- |
| `Temperature` | Daily maximum temperature (deg C) | 22 - 42 |
| `RH` | Relative humidity (%) | 21 - 90 |
| `Ws` | Wind speed (km/h) | 6 - 29 |
| `Rain` | Total rainfall for the day (mm) | 0 - 16.8 |

**Fire Weather Index system**

The six indices below are derived measures rather than direct readings, each built from
the weather columns above and, for the moisture codes, from the previous day's value.

| Column | Meaning | Range |
| --- | --- | --- |
| `FFMC` | Fine Fuel Moisture Code - dryness of surface litter | 28.6 - 92.5 |
| `DMC` | Duff Moisture Code - moisture in loosely packed organic layers | 1.1 - 65.9 |
| `DC` | Drought Code - moisture in deep, compact organic layers | 7 - 220.4 |
| `ISI` | Initial Spread Index - expected rate of fire spread | 0 - 18.5 |
| `BUI` | Buildup Index - total fuel available to burn | 1.1 - 68 |
| `FWI` | Fire Weather Index - overall fire intensity | 0 - 31.1 |

**Target**

| Column | Meaning |
| --- | --- |
| `Classes` | `fire` or `not fire` |
