## Lemmatization
Lemmatization also reduces a word to its base form, just like stemming. The difference is that the output (called a **lemma**) is always a **real dictionary word**.

Example: studies → study (stemming gives *studi*), went → go (stemming can't do this)

It is slower than stemming because it looks up words in a dictionary (WordNet), but the results are much more accurate. It is used in chatbots, Q&A systems and text summarization.

### Real life use case
**Customer support chatbot:** customers type *"my order is not arriving"*, *"order arrived late"*, *"when will my orders arrive?"*. Lemmatization turns arriving, arrived and arrive into **arrive**, and orders into **order**. Now the chatbot understands all these messages are about the same thing, delivery of an order, and gives the right reply.

```python
import nltk

# WordNet is the English dictionary the lemmatizer uses to find the base word
# only required once
nltk.download('wordnet')
```

### WordNet Lemmatizer

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

We can tell the lemmatizer what type of word it is (POS = Part of Speech):

| pos | meaning |
|---|---|
| `n` | noun (default) |
| `v` | verb |
| `a` | adjective |
| `r` | adverb |

```python
lemmatizer.lemmatize("going", pos="v")
```

```python
# same words we used in stemming, so we can compare
words = ["running", "runs", "ran", "playing", "played", "studies", "studying", "happiness", "university", "organization"]

for word in words:
    print(f"{word} ----> {lemmatizer.lemmatize(word, pos='v')}")
```

```python
# irregular verbs, stemming can't handle these
lemmatizer.lemmatize("went", pos="v"), lemmatizer.lemmatize("ate", pos="v"), lemmatizer.lemmatize("was", pos="v")
```

```python
# irregular nouns (pos is noun by default)
for word in ["children", "mice", "feet", "leaves"]:
    print(f"{word} ----> {lemmatizer.lemmatize(word)}")
```

```python
# adjectives
lemmatizer.lemmatize("better", pos="a"), lemmatizer.lemmatize("worse", pos="a")
```

### Why POS matters

```python
# without pos, 'was' is treated as a noun and gives a wrong result
lemmatizer.lemmatize("was"), lemmatizer.lemmatize("was", pos="v")
```

```python
# 'meeting' can be a noun (a meeting) or a verb (meeting someone)
lemmatizer.lemmatize("meeting", pos="n"), lemmatizer.lemmatize("meeting", pos="v")
```

### Stemming vs Lemmatization

```python
from nltk.stem import PorterStemmer

porter = PorterStemmer()

# stem can be a broken word, lemma is always a real word
for word in ["studies", "leaves", "flies", "geese"]:
    print(f"{word} ----> stem: {porter.stem(word)}, lemma: {lemmatizer.lemmatize(word)}")
```
