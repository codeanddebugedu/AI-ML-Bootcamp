```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Basic plotting with help of seaborn
# Search for seaborn dataset
tips=sns.load_dataset("tips")
tips
```

```python
sns.scatterplot(x="total_bill",y="tip",data=tips)
plt.show()
```

```python
# Using line plot
sns.lineplot(x="size",y="total_bill",data=tips)
plt.show()
```

```python
sns.barplot(x="day",y="total_bill",data=tips)
plt.show()
```

```python
# Box Plot
sns.boxplot(x="day", y="total_bill", data=tips)
plt.show()
```

```python
# Voilin Plot
sns.violinplot(x="day", y="total_bill", data=tips)
plt.show()
```

```python
# Histogram 
# total bill - lowest (5) highest(60)
# bin = 10
# 5 to 60 will get divided 10 parts
sns.histplot(tips["total_bill"], bins=10, kde=True) # Kernel Density
plt.show()
```

```python
# Standalone KDE Plot
sns.kdeplot(data=tips, x="total_bill", fill=True)
plt.show()
```

```python
sns.pairplot(tips)
plt.show()
```

```python
# Correlation
# x changes, tell me how y changes
# age - height (inc,inc)
# area, price (inc,inc)
# speed,time (inc,dec)
numeric_data=tips[["total_bill","tip","size"]]
correlation=numeric_data.corr()
correlation
```

```python
sns.heatmap(correlation,annot=True,cmap="coolwarm")
plt.show()
```
