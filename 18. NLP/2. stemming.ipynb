## Stemming
Stemming cuts a word down to its base form (stem) by removing endings like **-ing**, **-s**, **-ed**.

Example: running, runs → run

It helps in NLP tasks (like checking if a review is positive or negative) because different forms of the same word are treated as one word.


### Real life use case
**Search on a shopping website:** you type *"running shoes"*. Without stemming, the search looks only for the exact word "running" and misses products named *"Run Shoes"* or *"Shoes for runs"*. With stemming, running, run and runs all become **run**, so you see all the matching products.

**Product reviews:** a company wants to know what customers complain about. Reviews say *"delivery delayed"*, *"delays again"*, *"always delaying"*. Stemming turns all of them into **delay**, so the company can count them as one problem.

```python
# different forms of the same words
words = ["running", "runs", "ran", "playing", "played", "studies", "studying", "happiness", "university", "organization"]
```

### Porter Stemmer

```python
from nltk.stem import PorterStemmer

porter = PorterStemmer()

for word in words:
    print(f"{word} ----> {porter.stem(word)}")
```

```python
# stem is not always a real word
porter.stem("flies")
```

```python
# different meanings, same stem (a limitation of stemming)
porter.stem("universe"), porter.stem("university")
```

### Regexp Stemmer
Here we give our own pattern (regular expression). Any ending that matches the pattern gets removed from the word.

```python
from nltk.stem import RegexpStemmer

# remove 'ing', 's', 'e' or 'able' from the end ($ means end of word)
# min=4 -> words shorter than 4 letters are not stemmed
regex = RegexpStemmer("ing$|s$|e$|able$", min=4)
```

```python
# works well
for word in ["teaching", "books", "readable", "washable"]:
    print(f"{word} ----> {regex.stem(word)}")
```

```python
# goes wrong: it blindly cuts the ending without knowing the word
for word in ["king", "table", "care", "news", "notable", "running"]:
    print(f"{word} ----> {regex.stem(word)}")
```

```python
# 'bus' has only 3 letters, so min=4 leaves it as it is
regex.stem("bus")
```

```python
# only the ending 'ing' is removed, not the starting one
regex.stem("ingteaching")
```

### Snowball Stemmer
Snowball is an improved version of Porter (also called Porter2). It fixes some of Porter's mistakes and gives cleaner stems.

```python
from nltk.stem import SnowballStemmer

snowball = SnowballStemmer("english")

for word in words:
    print(f"{word} ----> {snowball.stem(word)}")
```

### Porter vs Snowball

```python
porter.stem("quickly"), porter.stem("kindly")
```

```python
# snowball handles 'ly' words better
snowball.stem("quickly"), snowball.stem("kindly")
```

```python
# both still make mistakes sometimes
porter.stem("goes"), snowball.stem("goes")
```
