```python
import matplotlib.pyplot as plt
```

```python
# Line Plot
x = [1, 2, 3, 4, 5, 6]
y = [1, 4, 9, 16, 25, 10]

plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("My Lineplot")
plt.show()
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [1, 4, 9, 16, 25, 10]

# Add some customization
plt.plot(x, y, color="green", linestyle="-.", marker="o", linewidth=2, markersize=10)
plt.grid(True)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [1, 2, 3, 4, 5]

plt.figure(figsize=(6, 3))
plt.subplot(2, 2, 1)
plt.plot(x, y1, color="red")
plt.title("Plot 1")

plt.subplot(2, 2, 2)
plt.plot(x, y2, color="green")
plt.title("Plot 2")

plt.subplot(2, 2, 3)
plt.plot(x, y1, color="brown")

plt.subplot(2, 2, 4)
plt.plot(x, y2, color="yellow")

plt.show()
```

```python
# Bar plot
x = ["A", "B", "C", "D", "E"]
y = [7, 4, 14, 12, 9]

plt.bar(x, y, color="red")
plt.show()
```

```python
# Histogram
data = [1, 2, 2, 3, 3, 3, 4, 4, 4, 20, 4, 5, 5, 5, 5, 5, 20]
plt.hist(data, bins=7, color="orange", edgecolor="black")
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 14, 2, 16, 25]
plt.scatter(x, y, color="blue", marker="x")
plt.show()
```

```python
l = ["cricket", "tennis", "tshirt", "football"]
sizes = [99, 45, 129, 68]
colors = ["orange", "skyblue", "green", "yellowgreen"]
ex = (0.1, 0.1, 0.1, 0.1)
plt.pie(sizes, labels=l, autopct="%1.1f%%", explode=ex, shadow=True)
plt.show()
```

```python
import pandas as pd

sales_data_df = pd.read_csv("sales_data_large.csv")
sales_data_df
```

```python
# Plot total sales by product
total_sales_by_product = sales_data_df.groupby("Product Category")["Units Sold"].sum()
total_sales_by_product
```

```python
total_sales_by_product.plot(kind="bar", color="teal")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.title("Total sales by product category")
plt.show()
```

```python
# Plot a sales trend by date
sales_trend = sales_data_df.groupby("Date")["Total Revenue"].sum().reset_index()
sales_trend
```

```python
plt.plot(sales_trend["Date"], sales_trend["Total Revenue"])
plt.show()
```
