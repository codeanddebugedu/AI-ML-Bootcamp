```python
from sklearn.datasets import make_classification

x,y = make_classification(n_samples=1000,n_features=2,n_clusters_per_class=1,weights=[0.90],random_state=16,n_redundant=0)
```

```python
x
```

```python
y
```

```python
import pandas as pd

df1=pd.DataFrame(x,columns=["amount","time"])
df2=pd.DataFrame(y,columns=["is_fraud"])

df = pd.concat([df1,df2],axis=1)
df
```

```python
df.is_fraud.value_counts()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['amount'],df['time'],c=df['is_fraud'])
```

```python
from imblearn.over_sampling import SMOTE

oversample=SMOTE()
x,y=oversample.fit_resample(df[['amount','time']],df["is_fraud"])
```
