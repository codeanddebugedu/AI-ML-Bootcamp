## Tokenization using NLTK

```python
import nltk

# download the tokenizer data (rules for splitting sentences and words)
# sent_tokenize and word_tokenize need it, only required once
nltk.download('punkt_tab')
```

### Sample text

```python
corpus = """Hi there, welcome to Code and Debug's AI-ML Bootcamp.
Learning NLP is fun! Let's start with tokenization today.
"""
print(corpus)
```

### Sentence Tokenization (paragraph → sentences)

```python
# splits text into sentences at . ! ? (full stop, exclamation, question mark)
# it is smart enough to not split on short forms like Mr. or Dr.
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(corpus)
print(sentences)
type(sentences)  # returns a list
```

```python
for sentence in sentences:
    print(sentence)
```

### Word Tokenization (paragraph / sentence → words)

```python
from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```

```python
# words of each sentence separately
for sentence in sentences:
    print(word_tokenize(sentence))
```

### WordPunct Tokenizer

```python
from nltk.tokenize import wordpunct_tokenize

wordpunct_tokenize(corpus)  # splits punctuation too, e.g. Debug ' s
```

### Treebank Tokenizer

```python
# Treebank tokenizer follows the rules used by the Penn Treebank (a famous English dataset)
# - splits short forms: Let's -> Let + 's, don't -> do + n't
# - separates punctuation like , ! ? into their own tokens
# - keeps hyphenated words together: AI-ML stays one token
# - full stop is split only at the very end, not in the middle
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
print(corpus)
tokenizer.tokenize(corpus)
```
